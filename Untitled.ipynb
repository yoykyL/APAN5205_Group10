```python
import pandas as pd
import os

folder = "."

dfs = []

for file in os.listdir(folder):
    
    if file.endswith(".csv"):
        
        skill = file.replace(".csv","")
        
        df = pd.read_csv(os.path.join(folder,file))
        
        df = df.rename(columns={
            df.columns[0]:"date",
            df.columns[1]:skill
        })
        
        dfs.append(df)
```

```python
import pandas as pd
from functools import reduce

cleaned = []
for i, df in enumerate(dfs):
    d = df.copy()

    if "date" not in d.columns:
        raise ValueError(f"dfs[{i}] 没有 date 列，当前列名: {d.columns.tolist()}")
    d["date"] = pd.to_datetime(d["date"])

    d = d.drop(columns=[c for c in d.columns if c.startswith("yhat")], errors="ignore")

    value_cols = [c for c in d.columns if c != "date"]
    if len(value_cols) == 0:
        raise ValueError(f"dfs[{i}] 除了 date 没有别的列了")
    if len(value_cols) > 1:

        value_col = value_cols[0]
    else:
        value_col = value_cols[0]

    new_name = f"skill_{i}"
    d = d[["date", value_col]].rename(columns={value_col: new_name})

    cleaned.append(d)

trends = reduce(lambda l, r: l.merge(r, on="date", how="outer"), cleaned)
trends = trends.sort_values("date").reset_index(drop=True)

trends.head()
```

```python
trends["date"] = pd.to_datetime(trends["date"])
trends = trends.sort_values("date")
trends.set_index("date", inplace=True)
```
